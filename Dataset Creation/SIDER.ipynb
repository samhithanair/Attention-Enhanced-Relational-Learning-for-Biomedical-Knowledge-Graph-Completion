```python
se_cols = ['drug_id', 'substance_id', 'side_effect_id', 'type', 'meddra_id', 'side_effect_name']
drug_cols = ['drug_id', 'drug_name']

side_effects = pd.read_csv(
    'C:\\Users\\ekru\\Documents\\Cursor\\ganv2\\Project\\SIDER\\meddra_all_se.tsv',
    sep='\t',
    names=se_cols,
    usecols=[0, 1, 2, 3, 4, 5]
)

drug_names = pd.read_csv(
    'C:\\Users\\ekru\\Documents\\Cursor\\ganv2\\Project\\SIDER\\drug_names.tsv',
    sep='\t',
    names=drug_cols
)
```

```python
#To check ODKG triplets
kg_triplets = pd.read_csv("C:\\Users\\ekru\\Documents\\Cursor\\ganv2\\Project\\Dataset Creation\\ODKG_triplets.csv")
openkb_drugs = set(kg_triplets['Head'].unique())
```

```python
#get unique drug-side effect relationships
unique_se = side_effects[side_effects['type'] == 'PT'].drop_duplicates(subset=['drug_id', 'side_effect_name'])
```

```python
#Creating databse 
drug_id_to_name = dict(zip(drug_names['drug_id'], drug_names['drug_name']))
se_relations = pd.merge(unique_se, drug_names, on='drug_id', how='left')
se_relations = se_relations[se_relations['drug_name'].isin(openkb_drugs)]
has_side_effect = pd.DataFrame({
    'Head': se_relations['drug_name'],
    'Relation': 'HAS_SIDE_EFFECT',
    'Tail': se_relations['side_effect_name']
})

all_relations = pd.concat([has_side_effect])
```

```python
# Save SIDER relationships to a separate file
all_relations.to_csv('SIDER_triplets.csv', index=False)
print(all_relations.head())
```
