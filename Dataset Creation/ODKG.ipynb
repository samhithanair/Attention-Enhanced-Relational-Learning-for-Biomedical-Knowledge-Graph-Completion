```python
import pandas as pd
```

```python
condition = pd.read_csv('C:\\Users\\ekru\\Documents\\Cursor\\ganv2\\Project\\ODKG\\condition.csv')
drug = pd.read_csv('C:\\Users\\ekru\\Documents\\Cursor\\ganv2\\Project\\ODKG\\drug.csv')
interaction = pd.read_csv('C:\\Users\\ekru\\Documents\\Cursor\\ganv2\\Project\\ODKG\\interaction.csv')
manufacturer = pd.read_csv('C:\\Users\\ekru\\Documents\\Cursor\\ganv2\\Project\\ODKG\\manufacturer.csv')
price = pd.read_csv('C:\\Users\\ekru\\Documents\\Cursor\\ganv2\\Project\\ODKG\\price.csv')
product = pd.read_csv('C:\\Users\\ekru\\Documents\\Cursor\\ganv2\\Project\\ODKG\\product.csv')
store = pd.read_csv('C:\\Users\\ekru\\Documents\\Cursor\\ganv2\\Project\\ODKG\\store.csv')
treatment = pd.read_csv('C:\\Users\\ekru\\Documents\\Cursor\\ganv2\\Project\\ODKG\\treatment.csv')
```

```python
triplets = []
```

```python
# (Condition) -[TREATED_BY]-> (Drug)
for _, row in treatment.iterrows():
    condition_name = condition.loc[condition['id'] == row['condition_id'], 'name'].values[0]
    drug_name = drug.loc[drug['id'] == row['drug_id'], 'name'].values[0]
    triplets.append((condition_name, 'TREATED_BY', drug_name))
```

```python
# (Drug) -[IN_PRODUCT]-> (Product)
for _, row in product.iterrows():
    product_name = row['name']
    drug_name = drug.loc[drug['id'] == row['drug_id'], 'name'].values[0]
    triplets.append((drug_name, 'IN_PRODUCT', product_name))
```

```python
# (Product) -[MANUFACTURED_BY]-> (Manufacturer)
for _, row in product.iterrows():
    product_name = row['name']
    manufacturer_name = manufacturer.loc[manufacturer['id'] == row['manufacturer_id'], 'name'].values[0] if pd.notna(row['manufacturer_id']) else None
    if manufacturer_name:
        triplets.append((product_name, 'MANUFACTURED_BY', manufacturer_name))
```

```python
# (Product) -[HAS_PRICE]-> (Price)
for _, row in price.iterrows():
    product_name = product.loc[product['id'] == row['product_id'], 'name'].values[0]
    price_value = row['price']
    triplets.append((product_name, 'HAS_PRICE', price_value))
```

```python
# (Price) -[AVAILABLE_AT]-> (Store)
for _, row in price.iterrows():
    price_value = row['price']
    store_name = store.loc[store['id'] == row['store_id'], 'name'].values[0]
    triplets.append((price_value, 'AVAILABLE_AT', store_name))
```

```python
# (Drug) -[INTERACTS_WITH]-> (Drug)
for _, row in interaction.iterrows():
    source_drug = drug.loc[drug['id'] == row['source_drug_id'], 'name'].values[0]
    target_drug = drug.loc[drug['id'] == row['target_drug_id'], 'name'].values[0]
    triplets.append((source_drug, 'INTERACTS_WITH', target_drug))
```

```python
ODKG_df = pd.DataFrame(triplets, columns=['Head', 'Relation', 'Tail'])
ODKG_df = ODKG_df.sample(frac=1).reset_index(drop=True)
ODKG_df.to_csv('ODKG_triplets.csv', index=False)
```
